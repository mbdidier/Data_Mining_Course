
## Exercise 4: Real-life Applications of Statistical Learning

This notebook presents clean and polished answers to Exercise 4, covering **classification**, **regression**, and **cluster analysis**, with real-life examples.


## (a) Classification

Classification is used when the response variable is **categorical**.


**1. Plant classification (weed vs useful plant)**  
The goal is to predict whether a plant shown in an image is a weed or a useful plant.

- **Response:** Weed / Useful plant  
- **Predictors:** Color, shape, texture, and other visual features extracted from the image  
- **Goal:** Prediction  

This is a classification problem because the response takes one of two categories.


**2. Spam detection**  
The goal is to classify an email or message as spam or not spam.

- **Response:** Spam / Not spam  
- **Predictors:** Frequency of certain words, presence of links, message length, sender’s email or domain  
- **Goal:** Prediction  

The focus is on accurately classifying new incoming messages.


**3. Startup success analysis**  
The goal is to understand which factors influence whether a startup company succeeds or fails.

- **Response:** Success / Failure  
- **Predictors:** Market sector, business model, company size, amount of investment  
- **Goal:** Inference  

Although this is a classification problem, the primary interest is understanding which predictors most affect success.


## (b) Regression

Regression is used when the response variable is **quantitative**.


**1. Marketing analysis**  
The goal is to understand how different advertising methods affect sales.

- **Response:** Number of sales  
- **Predictors:** Spending on different advertising channels (TV, online, radio, etc.)  
- **Goal:** Inference  

The objective is to identify which advertising methods have the strongest impact on sales.


**2. Sales forecasting**  
The goal is to predict the number of sales at the end of a future time period.

- **Response:** Number of sales  
- **Predictors:** Past sales, seasonality, promotions, economic factors  
- **Goal:** Prediction  

This helps businesses plan inventory and operations.


**3. House price forecasting**  
The goal is to predict the price of a house based on its characteristics.

- **Response:** House price  
- **Predictors:** Location, number of rooms, construction date, house size  
- **Goal:** Prediction  

This is a standard regression problem with a numeric response.


## (c) Cluster Analysis

Cluster analysis is an **unsupervised** learning method where there is no response variable.


**1. Disease grouping**  
Diseases can be grouped based on shared characteristics.

- **Features:** Symptoms, genetic markers, laboratory test results  
- **Usefulness:** Helps identify disease subtypes and improve diagnosis or treatment strategies.


**2. User clustering for recommendation systems**  
Users can be grouped based on their media consumption behavior.

- **Features:** Viewing history, ratings, frequency of use  
- **Usefulness:** Clusters can be used to generate personalized recommendations.


**3. Anomaly detection**  
Clustering can help identify unusual observations in data.

- **Features:** Measurements or attributes of normal observations  
- **Usefulness:** Data points that do not belong to any cluster or are far from others may indicate anomalies or errors.


## Summary

- **Classification:** categorical response variable  
- **Regression:** numeric response variable  
- **Clustering:** no response variable, used to discover structure in data
