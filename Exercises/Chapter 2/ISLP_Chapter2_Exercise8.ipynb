## Exercise 8: College Data Set (Exploratory Data Analysis)

This notebook provides a complete solution to Exercise 8 using the **College** data set.


## (a) Load the data

We read the `College.csv` file into Python using `pandas`.


```python
import pandas as pd
import matplotlib.pyplot as plt

college = pd.read_csv('College.csv')
college.head()
```

## (b) Fix the first column (college names)

The first column contains the names of the colleges and should be used as the index.


```python
college2 = pd.read_csv('College.csv', index_col=0)
college3 = college.rename({'Unnamed: 0': 'College'}, axis=1)
college3 = college3.set_index('College')

college = college3
college.head()
```

## (c) Numerical summary of the data

We use the `describe()` method to obtain summary statistics.


```python
college.describe()
```

## (d) Scatterplot matrix

We create a scatterplot matrix for `Top10perc`, `Apps`, and `Enroll`.


```python
pd.plotting.scatter_matrix(
    college[['Top10perc', 'Apps', 'Enroll']],
    figsize=(8, 8)
)
plt.show()
```

## (e) Boxplots of Out-of-state tuition vs Private


```python
college.boxplot(column='Outstate', by='Private', figsize=(6, 4))
plt.title('Out-of-State Tuition by Private/Public')
plt.suptitle('')
plt.xlabel('Private')
plt.ylabel('Out-of-State Tuition')
plt.show()
```

## (f) Create an Elite variable


```python
college['Elite'] = pd.cut(
    college['Top10perc'],
    [0, 50, 100],
    labels=['No', 'Yes']
)

college['Elite'].value_counts()
```

### Boxplots of Out-of-state tuition vs Elite


```python
college.boxplot(column='Outstate', by='Elite', figsize=(6, 4))
plt.title('Out-of-State Tuition by Elite Status')
plt.suptitle('')
plt.xlabel('Elite')
plt.ylabel('Out-of-State Tuition')
plt.show()
```

## (g) Histograms of quantitative variables


```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

college['Apps'].plot.hist(bins=20, ax=axes[0, 0])
axes[0, 0].set_title('Applications')

college['Enroll'].plot.hist(bins=20, ax=axes[0, 1])
axes[0, 1].set_title('Enrollment')

college['Outstate'].plot.hist(bins=30, ax=axes[1, 0])
axes[1, 0].set_title('Out-of-State Tuition')

college['Grad.Rate'].plot.hist(bins=15, ax=axes[1, 1])
axes[1, 1].set_title('Graduation Rate')

plt.tight_layout()
plt.show()
```

## (h) Summary of findings

- Private colleges generally charge higher out-of-state tuition than public colleges.
- Elite colleges tend to have significantly higher out-of-state tuition than non-elite colleges.
- Applications and enrollment numbers are highly skewed.
- Graduation rates vary, with a few unusually high or low values.
