# Linear Regression Without an Intercept — Exercise 5

## Question
In linear regression **without an intercept**, the fitted value is

$$
\hat{y}_i = x_i\,\hat{\beta}
$$

where

$$
\hat{\beta} = \frac{\sum_{i'=1}^{n} x_{i'} y_{i'}}{\sum_{i'=1}^{n} x_{i'}^{2}}.
$$

Show that

$$
\hat{y}_i = \sum_{i'=1}^{n} a_{i'} y_{i'}
$$

and find $a_{i'}$.



## Simple step-by-step solution

### Step 1: Start with the fitted value
$$
\hat{y}_i = x_i\hat{\beta}
$$

### Step 2: Substitute the formula for $\hat{\beta}$ (replace it)
$$
\hat{y}_i = x_i\left(\frac{\sum_{i'=1}^{n} x_{i'} y_{i'}}{\sum_{i'=1}^{n} x_{i'}^{2}}\right)
$$

### Step 3: Pull out constants (they don’t depend on $i'$)
$$
\hat{y}_i = \frac{x_i}{\sum_{k=1}^{n} x_k^{2}}\left(\sum_{i'=1}^{n} x_{i'}y_{i'}\right)
$$

### Step 4: Distribute the constant into the sum
$$
\hat{y}_i = \sum_{i'=1}^{n}\left(\frac{x_i x_{i'}}{\sum_{k=1}^{n} x_k^{2}}\right)y_{i'}
$$

### Step 5: Read off the coefficient (weight)
Comparing with
$$
\hat{y}_i = \sum_{i'=1}^{n} a_{i'}y_{i'},
$$
we get

$$
\boxed{a_{i'} = \frac{x_i x_{i'}}{\sum_{k=1}^{n} x_k^{2}}.}
$$

> Note: the weight depends on **both** $i$ and $i'$. Some books write $a_{ii'}$ instead of $a_{i'}$.



## Quick numeric example (to make it feel real)

Let’s pick 3 data points:

- $x = [1, 2, 3]$
- $y = [2, 1, 4]$

We compute $\hat\beta$, then the fitted values $\hat y$, and then verify that each $\hat y_i$ is a weighted sum of $y_1,y_2,y_3$.



```python
import numpy as np

x = np.array([1, 2, 3], dtype=float)
y = np.array([2, 1, 4], dtype=float)
n = len(x)

beta_hat = (x @ y) / (x @ x)   # sum(x*y) / sum(x^2)
y_hat = x * beta_hat

beta_hat, y_hat

```

### Build the weights and check the formula

From the exercise, the weights for a fixed $i$ are:

$$
a_{i'} = \frac{x_i x_{i'}}{\sum_{k=1}^n x_k^2}.
$$

So for each $i$, we compute all weights and check:

$$
\hat y_i = a_1y_1 + a_2y_2 + \dots + a_ny_n.
$$



```python
den = x @ x  # sum of squares

for i in range(n):
    a = (x[i] * x) / den            # weights for this i (vector of length n)
    y_hat_from_weights = a @ y      # sum(a_i' * y_i')
    print(f"i={i+1}: weights={a},  weighted-sum={y_hat_from_weights:.6f},  direct={y_hat[i]:.6f}")

```

If the two numbers match for each $i$, then the exercise is verified.
