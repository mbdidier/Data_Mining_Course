# Exercise 12: Logistic Regression vs Softmax Formulation

## Problem Statement

Suppose that you wish to classify an observation $X \in \mathbb{R}$ into **apples** and **oranges**.

**Your Model (Standard Logistic Regression):**
$$\hat{Pr}(Y = \text{orange} | X = x) = \frac{\exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}$$

**Friend's Model (Softmax Formulation):**
$$\hat{Pr}(Y = \text{orange} | X = x) = \frac{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1} x)}{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1} x) + \exp(\hat{\alpha}_{\text{apple}0} + \hat{\alpha}_{\text{apple}1} x)}$$

```python
import numpy as np
import sympy as sp
from sympy import symbols, exp, log, simplify, Eq

print("Libraries imported successfully!")
```

## Part (a): Log Odds of Orange vs Apple in YOUR Model

**Question:** What is the log odds of orange versus apple in your model?

### Solution:

In the standard logistic regression model:

\begin{align*}
log(\frac{Pr(Y = orange | X = x)}{Pr(Y = apple | X = x)}) &= log(\frac{Pr(Y = orange | X = x)}{1 - Pr(Y = orange | X = x)}) \\ 

&= log(\frac{\frac{exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + exp(\hat{\beta}_0 + \hat{\beta}_1 x)}}{1 - \frac{exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + exp(\hat{\beta}_0 + \hat{\beta}_1 x)}}) \\

&= log(\frac{\frac{exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + exp(\hat{\beta}_0 + \hat{\beta}_1 x)}}{\frac{1}{1 + exp(\hat{\beta}_0 + \hat{\beta}_1 x)}})  \\ 

&= log(exp(\hat{\beta}_0 + \hat{\beta}_1 x)) \\

&= \hat{\beta}_0 + \hat{\beta}_1 x

\end{align*}

The log odds of orange versus apple in my model is $\hat{\beta}_0 + \hat{\beta}_1 x$

### Solution:

In the standard logistic regression model:

\begin{align*}
log(\frac{Pr(Y = orange | X = x)}{Pr(Y = apple | X = x)}) &= log(\frac{Pr(Y = orange | X = x)}{1 - Pr(Y = orange | X = x)}) \\ 

&= log(\frac{\frac{exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + exp(\hat{\beta}_0 + \hat{\beta}_1 x)}}{1 - \frac{exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + exp(\hat{\beta}_0 + \hat{\beta}_1 x)}}) \\

&= log(\frac{\frac{exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + exp(\hat{\beta}_0 + \hat{\beta}_1 x)}}{\frac{1}{1 + exp(\hat{\beta}_0 + \hat{\beta}_1 x)}})  \\ 

&= log(exp(\hat{\beta}_0 + \hat{\beta}_1 x)) \\

&= \hat{\beta}_0 + \hat{\beta}_1 x

\end{align*}

The log odds of orange versus apple in my model is $\hat{\beta}_0 + \hat{\beta}_1 x$

```python
# Define symbols
beta_0, beta_1, x = symbols('beta_0 beta_1 x')

# Your model
Pr_orange_your = exp(beta_0 + beta_1 * x) / (1 + exp(beta_0 + beta_1 * x))
Pr_apple_your = 1 / (1 + exp(beta_0 + beta_1 * x))

# Log odds
log_odds_your = log(Pr_orange_your / Pr_apple_your)
log_odds_your_simplified = simplify(log_odds_your)

print("YOUR MODEL (Standard Logistic Regression)")
print("="*50)
print(f"\nPr(orange|X=x) = exp(β₀ + β₁x) / (1 + exp(β₀ + β₁x))")
print(f"\nPr(apple|X=x)  = 1 / (1 + exp(β₀ + β₁x))")
print(f"\nLog Odds = log[Pr(orange) / Pr(apple)]")
print(f"\n         = log[exp(β₀ + β₁x) / 1]")
print(f"\n         = β₀ + β₁x")
```

### Answer (a):

$$\boxed{\log\left(\frac{Pr(\text{orange})}{Pr(\text{apple})}\right) = \hat{\beta}_0 + \hat{\beta}_1 x}$$

## Part (b): Log Odds of Orange vs Apple in FRIEND'S Model

**Question:** What is the log odds of orange versus apple in your friend's model?

### Solution:

In the softmax formulation:

\begin{align*}

log(\frac{Pr(Y = orange | X = x)}{Pr(Y = apple | X = x)}) &= log(\frac{\frac{exp(\hat{\alpha}_{orange0} + \hat{\alpha}_{orange1} x)}{exp(\hat{\alpha}_{orange0} + \hat{\alpha}_{orange1} x) + exp(\hat{\alpha}_{apple0} + \hat{\alpha}_{apple1} x)}}{\frac{exp(\hat{\alpha}_{apple0} + \hat{\alpha}_{apple1} x)}{exp(\hat{\alpha}_{orange0} + \hat{\alpha}_{orange1} x) + exp(\hat{\alpha}_{apple0} + \hat{\alpha}_{apple1} x)}}) \\ 

&= log(\frac{exp(\hat{\alpha}_{orange0} + \hat{\alpha}_{orange1} x)}{exp(\hat{\alpha}_{apple0} + \hat{\alpha}_{apple1} x)}) \\ 

&= (\hat{\alpha}_{orange0} - \hat{\alpha}_{apple0}) + (\hat{\alpha}_{orange1} - \hat{\alpha}_{apple1}) x\\

\end{align*}

The log odds of orange versus apple in our friend's model is $(\hat{\alpha}_{orange0} - \hat{\alpha}_{apple0}) + (\hat{\alpha}_{orange1} - \hat{\alpha}_{apple1}) x$

```python
# Define symbols for friend's model
a_or0, a_or1, a_ap0, a_ap1 = symbols('alpha_orange0 alpha_orange1 alpha_apple0 alpha_apple1')

# Friend's model (softmax)
numerator_orange = exp(a_or0 + a_or1 * x)
numerator_apple = exp(a_ap0 + a_ap1 * x)
denominator = numerator_orange + numerator_apple

Pr_orange_friend = numerator_orange / denominator
Pr_apple_friend = numerator_apple / denominator

# Log odds
log_odds_friend = log(Pr_orange_friend / Pr_apple_friend)
log_odds_friend_simplified = simplify(log_odds_friend)

print("FRIEND'S MODEL (Softmax Formulation)")
print("="*50)
print(f"\nPr(orange|X=x) = exp(α_or0 + α_or1·x) / [exp(α_or0 + α_or1·x) + exp(α_ap0 + α_ap1·x)]")
print(f"\nPr(apple|X=x)  = exp(α_ap0 + α_ap1·x) / [exp(α_or0 + α_or1·x) + exp(α_ap0 + α_ap1·x)]")
print(f"\nLog Odds = log[Pr(orange) / Pr(apple)]")
print(f"\n         = log[exp(α_or0 + α_or1·x) / exp(α_ap0 + α_ap1·x)]")
print(f"\n         = (α_or0 + α_or1·x) - (α_ap0 + α_ap1·x)")
print(f"\n         = (α_or0 - α_ap0) + (α_or1 - α_ap1)·x")
```

### Answer (b):

$$\boxed{\log\left(\frac{Pr(\text{orange})}{Pr(\text{apple})}\right) = (\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0}) + (\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1}) x}$$

## Part (c): Convert Your Coefficients to Friend's Coefficients

**Question:** Suppose that in your model, $\hat{\beta}_0 = 2$ and $\hat{\beta}_1 = -1$. What are the coefficient estimates in your friend's model?

### Solution:

From parts (a) and (b), both models must give the same log odds:

$$\beta_0 + \beta_1 x = (\alpha_{\text{orange}0} - \alpha_{\text{apple}0}) + (\alpha_{\text{orange}1} - \alpha_{\text{apple}1}) x$$

Matching coefficients:
- $\beta_0 = \alpha_{\text{orange}0} - \alpha_{\text{apple}0}$
- $\beta_1 = \alpha_{\text{orange}1} - \alpha_{\text{apple}1}$

**Key Insight:** The softmax formulation is **overparameterized**. We only know the *differences* between coefficients, not their absolute values. There are infinitely many valid solutions!

```python
# Given values
beta_0_val = 2
beta_1_val = -1

print("CONVERTING YOUR COEFFICIENTS TO FRIEND'S MODEL")
print("="*60)
print(f"\nYour coefficients: β₀ = {beta_0_val}, β₁ = {beta_1_val}")
print(f"\nConstraints:")
print(f"  α_orange0 - α_apple0 = β₀ = {beta_0_val}")
print(f"  α_orange1 - α_apple1 = β₁ = {beta_1_val}")
print(f"\nThe softmax formulation is OVERPARAMETERIZED!")
print(f"We can only determine the DIFFERENCES, not absolute values.")
```

```python
# Show some valid solutions
print("\nExamples of Valid Coefficient Sets for Friend's Model:")
print("-"*60)

# Example 1: Set apple coefficients to 0
print(f"\nExample 1 (Set α_apple = 0):")
print(f"  α_orange0 = {beta_0_val}, α_orange1 = {beta_1_val}")
print(f"  α_apple0  = 0, α_apple1  = 0")

# Example 2: Set orange coefficients to 0
print(f"\nExample 2 (Set α_orange = 0):")
print(f"  α_orange0 = 0, α_orange1 = 0")
print(f"  α_apple0  = {-beta_0_val}, α_apple1  = {-beta_1_val}")

# Example 3: Split evenly
print(f"\nExample 3 (Split evenly):")
print(f"  α_orange0 = {beta_0_val/2}, α_orange1 = {beta_1_val/2}")
print(f"  α_apple0  = {-beta_0_val/2}, α_apple1  = {-beta_1_val/2}")

# Example 4: Arbitrary values
print(f"\nExample 4 (Arbitrary - add 5 to all):")
print(f"  α_orange0 = {beta_0_val + 5}, α_orange1 = {beta_1_val + 3}")
print(f"  α_apple0  = 5, α_apple1  = 3")
```

```python
# Verify all solutions give same probabilities
def calc_prob_orange_softmax(x_val, a_or0, a_or1, a_ap0, a_ap1):
    """Calculate P(orange|x) using softmax"""
    num = np.exp(a_or0 + a_or1 * x_val)
    denom = np.exp(a_or0 + a_or1 * x_val) + np.exp(a_ap0 + a_ap1 * x_val)
    return num / denom

def calc_prob_orange_logistic(x_val, b0, b1):
    """Calculate P(orange|x) using standard logistic"""
    return np.exp(b0 + b1 * x_val) / (1 + np.exp(b0 + b1 * x_val))

# Test with x = 1
x_test = 1
print(f"\nVerification (all should give same probability for x = {x_test}):")
print("-"*60)

prob_logistic = calc_prob_orange_logistic(x_test, beta_0_val, beta_1_val)
print(f"Your model:         P(orange|x={x_test}) = {prob_logistic:.6f}")

prob_ex1 = calc_prob_orange_softmax(x_test, 2, -1, 0, 0)
print(f"Friend's Example 1: P(orange|x={x_test}) = {prob_ex1:.6f}")

prob_ex2 = calc_prob_orange_softmax(x_test, 0, 0, -2, 1)
print(f"Friend's Example 2: P(orange|x={x_test}) = {prob_ex2:.6f}")

prob_ex3 = calc_prob_orange_softmax(x_test, 1, -0.5, -1, 0.5)
print(f"Friend's Example 3: P(orange|x={x_test}) = {prob_ex3:.6f}")

prob_ex4 = calc_prob_orange_softmax(x_test, 7, 2, 5, 3)
print(f"Friend's Example 4: P(orange|x={x_test}) = {prob_ex4:.6f}")

print(f"\n✓ All probabilities are identical!")
```

### Answer (c):

The softmax formulation is **overparameterized**, so we cannot uniquely determine the coefficients. We can only say:

$$\boxed{\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0} = 2}$$
$$\boxed{\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1} = -1}$$

Any coefficients satisfying these constraints are valid. For example:
- $\hat{\alpha}_{\text{orange}0} = 2, \hat{\alpha}_{\text{orange}1} = -1, \hat{\alpha}_{\text{apple}0} = 0, \hat{\alpha}_{\text{apple}1} = 0$

## Part (d): Convert Friend's Coefficients to Your Coefficients

**Question:** Your friend gets: $\hat{\alpha}_{\text{orange}0} = 1.2$, $\hat{\alpha}_{\text{orange}1} = -2$, $\hat{\alpha}_{\text{apple}0} = 3$, $\hat{\alpha}_{\text{apple}1} = 0.6$. What are the coefficient estimates in your model?

```python
# Friend's coefficients
alpha_orange0 = 1.2
alpha_orange1 = -2
alpha_apple0 = 3
alpha_apple1 = 0.6

# Convert to your model
beta_0_new = alpha_orange0 - alpha_apple0
beta_1_new = alpha_orange1 - alpha_apple1

print("CONVERTING FRIEND'S COEFFICIENTS TO YOUR MODEL")
print("="*60)
print(f"\nFriend's coefficients:")
print(f"  α_orange0 = {alpha_orange0}")
print(f"  α_orange1 = {alpha_orange1}")
print(f"  α_apple0  = {alpha_apple0}")
print(f"  α_apple1  = {alpha_apple1}")

print(f"\nConversion:")
print(f"  β₀ = α_orange0 - α_apple0 = {alpha_orange0} - {alpha_apple0} = {beta_0_new}")
print(f"  β₁ = α_orange1 - α_apple1 = {alpha_orange1} - {alpha_apple1} = {beta_1_new}")
```

```python
# Verify both models give same probabilities
x_test_values = [-2, -1, 0, 1, 2, 3]

print("\nVerification - Both models give identical probabilities:")
print("-"*60)
print(f"{'x':<8} {'Your Model':<20} {'Friend Model':<20} {'Match?':<10}")
print("-"*60)

for x_val in x_test_values:
    prob_your = calc_prob_orange_logistic(x_val, beta_0_new, beta_1_new)
    prob_friend = calc_prob_orange_softmax(x_val, alpha_orange0, alpha_orange1, 
                                           alpha_apple0, alpha_apple1)
    match = "✓" if np.isclose(prob_your, prob_friend) else "✗"
    print(f"{x_val:<8} {prob_your:<20.6f} {prob_friend:<20.6f} {match:<10}")
```

### Answer (d):

$$\boxed{\hat{\beta}_0 = \hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0} = 1.2 - 3 = -1.8}$$
$$\boxed{\hat{\beta}_1 = \hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1} = -2 - 0.6 = -2.6}$$

## Part (e): Agreement Between Models on Test Data

**Question:** If you apply both models from (d) to a data set with 2,000 test observations, what fraction of the time do you expect the predicted class labels from your model to agree with those from your friend's model?

```python
print("AGREEMENT BETWEEN MODELS")
print("="*60)

print("""
Key Insight:
────────────
Both models compute EXACTLY the same probabilities for each class!

From parts (a) and (b), we showed that:
• Your model: log odds = β₀ + β₁x
• Friend's model: log odds = (α_or0 - α_ap0) + (α_or1 - α_ap1)x

When β₀ = α_or0 - α_ap0 and β₁ = α_or1 - α_ap1,
both models have IDENTICAL log odds.

Same log odds → Same probabilities → Same predictions!
""")

print("\nPrediction Rule (same for both models):")
print("-"*50)
print("  Predict 'orange' if P(orange|x) > 0.5")
print("  Predict 'apple'  if P(orange|x) ≤ 0.5")
print("\nSince both models produce IDENTICAL probabilities,")
print("they will ALWAYS make the same prediction!")
```

```python
# Demonstrate with 2000 random test observations
np.random.seed(42)
n_test = 2000

# Generate random x values
X_test = np.random.uniform(-5, 5, n_test)

# Predictions from your model
prob_your_model = calc_prob_orange_logistic(X_test, beta_0_new, beta_1_new)
pred_your_model = (prob_your_model > 0.5).astype(int)  # 1 = orange, 0 = apple

# Predictions from friend's model
prob_friend_model = calc_prob_orange_softmax(X_test, alpha_orange0, alpha_orange1,
                                              alpha_apple0, alpha_apple1)
pred_friend_model = (prob_friend_model > 0.5).astype(int)

# Calculate agreement
agreement = np.mean(pred_your_model == pred_friend_model)

print(f"\nSimulation with {n_test} Test Observations:")
print("="*60)
print(f"\nPredictions from Your Model:   {np.sum(pred_your_model == 1)} oranges, {np.sum(pred_your_model == 0)} apples")
print(f"Predictions from Friend's Model: {np.sum(pred_friend_model == 1)} oranges, {np.sum(pred_friend_model == 0)} apples")
print(f"\nNumber of agreements: {np.sum(pred_your_model == pred_friend_model)}")
print(f"Number of disagreements: {np.sum(pred_your_model != pred_friend_model)}")
print(f"\nFraction of agreement: {agreement:.2%}")
```

```python
# Verify probabilities are identical
prob_diff = np.abs(prob_your_model - prob_friend_model)
print(f"\nVerification - Probability Differences:")
print("-"*50)
print(f"Maximum difference: {np.max(prob_diff):.2e}")
print(f"Mean difference: {np.mean(prob_diff):.2e}")
print(f"\nThe tiny differences are due to floating-point precision,")
print(f"not actual model differences.")
```

### Answer (e):

$$\boxed{\text{The predicted class labels will agree 100\% of the time.}}$$

**Explanation:**

Both models are mathematically equivalent! They compute the exact same probabilities for each observation because:

1. The log odds are identical: $\beta_0 + \beta_1 x = (\alpha_{\text{orange}0} - \alpha_{\text{apple}0}) + (\alpha_{\text{orange}1} - \alpha_{\text{apple}1})x$

2. Same log odds → Same probabilities

3. Same probabilities → Same predictions (since both use the rule: predict "orange" if P(orange) > 0.5)

The standard logistic regression and the softmax formulation are simply two different **parameterizations** of the same model for binary classification.
